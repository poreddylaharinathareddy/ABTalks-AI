```python

from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

print("Sentence embedding model loaded successfully! 🚀")
```

```python

sentences = [
    "I love learning artificial intelligence.",
    "I enjoy studying AI and machine learning.",
    "The weather is sunny today.",
    "Python is a programming language.",
    "I like building AI applications."
]

print("Our five sentences:")
for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")
```

```python

embeddings = model.encode(sentences)

print("Sentence vectors generated successfully! 🚀")
print("Number of sentences:", len(embeddings))
print("Vector size:", embeddings.shape[1])

print("\nFirst sentence vector (first 10 numbers):")
print(embeddings[0][:10])
```

```python

from sentence_transformers import util

pairs = [
    (0, 1),
    (0, 2),
    (0, 4),
    (1, 3),
    (3, 4)
]

print("Cosine Similarity Results:\n")

for i, j in pairs:
    score = util.cos_sim(embeddings[i], embeddings[j]).item()
    print(f"Sentence {i+1} vs Sentence {j+1}: {score:.4f}")
    print(f"  {sentences[i]}")
    print(f"  {sentences[j]}\n")
```

```python

from sentence_transformers import util

similarity_matrix = util.cos_sim(embeddings, embeddings)

print("Sentence Similarity Matrix:\n")
print(similarity_matrix.round(decimals=2))

# Find the most similar pair of different sentences
best_score = -1
best_pair = None

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        score = similarity_matrix[i][j].item()
        if score > best_score:
            best_score = score
            best_pair = (i, j)

print("\nMost similar sentence pair:")
print(sentences[best_pair[0]])
print(sentences[best_pair[1]])
print("Similarity score:", round(best_score, 4))
```
