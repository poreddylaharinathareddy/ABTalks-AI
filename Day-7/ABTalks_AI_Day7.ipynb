```python

# ABTalks AI Challenge - Day 7
# Sentiment Classification

print("Day 7: Sentiment Classification")
print("Learning how AI identifies positive and negative reviews!")
```

```python

# Step 2: Create a small sentiment dataset

import pandas as pd

data = {
    "review": [
        "I love this product",
        "This movie is amazing",
        "The food was delicious",
        "I am very happy",
        "Excellent service",
        "I hate this product",
        "This movie is terrible",
        "The food was awful",
        "I am very disappointed",
        "Poor service"
    ],
    "sentiment": [
        "positive", "positive", "positive", "positive", "positive",
        "negative", "negative", "negative", "negative", "negative"
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python

# Step 3: Train a sentiment classification model

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Convert text reviews into numbers
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df["review"])

# Prepare sentiment labels
y = df["sentiment"]

# Train the model
model = MultinomialNB()
model.fit(X, y)

print("Sentiment model trained successfully!")
```

```python

# Step 4: Evaluate model accuracy

from sklearn.metrics import accuracy_score

# Predict sentiments for our training reviews
predictions = model.predict(X)

accuracy = accuracy_score(y, predictions)

print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python

# Step 5: Test the model with 5 new reviews

test_reviews = [
    "The product is wonderful",
    "I really enjoyed this movie",
    "The service was excellent",
    "I dislike this product",
    "The food was horrible"
]

# Convert the new reviews into numbers
test_vectors = vectorizer.transform(test_reviews)

# Predict sentiment
test_predictions = model.predict(test_vectors)

# Display results
for review, sentiment in zip(test_reviews, test_predictions):
    print(f"Review: {review}")
    print(f"Predicted sentiment: {sentiment}")
    print("-" * 40)
```

```python

# Step 6: Summary of the sentiment classification project

print("DAY 7: SENTIMENT CLASSIFICATION")
print("=" * 40)

print("Training reviews:", len(df))
print("Positive reviews:", (df["sentiment"] == "positive").sum())
print("Negative reviews:", (df["sentiment"] == "negative").sum())

print("\nModel accuracy on training data:",
      round(accuracy * 100, 2), "%")

print("\nKey Learnings:")
print("1. Text can be converted into numerical features.")
print("2. Naive Bayes can classify positive and negative reviews.")
print("3. Small datasets may not generalize well to new reviews.")
print("4. More varied training examples can improve the model.")
```
