# Day 9: Vector Representation and Cosine Similarity

## Objective
To convert text into numerical vectors using TF-IDF, calculate cosine similarity between sentences, build a semantic similarity search function, and visualize similarity scores using a heatmap.

## Introduction
Computers cannot directly understand or compare plain text mathematically. Therefore, text must be converted into numerical representations called vectors.

**TF-IDF (Term Frequency–Inverse Document Frequency)** represents text based on the importance of words in a collection of documents.

**Cosine similarity** measures how similar two vectors are by calculating the cosine of the angle between them. A higher score generally indicates greater similarity in their word patterns.

```python
# Import required libraries

import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python
# Create a corpus of 10 sentences covering 3 topics

corpus = [
    # Topic 1: Animals
    "The dog plays in the garden",
    "A puppy is a young dog",
    "Cats are friendly domestic animals",

    # Topic 2: Technology
    "Artificial intelligence helps computers learn",
    "Machine learning is a branch of artificial intelligence",
    "Computers process data using algorithms",

    # Topic 3: Cooking
    "I enjoy cooking delicious food",
    "Fresh vegetables make healthy meals",
    "Cooking pasta is easy and enjoyable",
    "A chef prepares tasty meals in the kitchen"
]

print("Total sentences:", len(corpus))
print("\nOur sentence corpus:\n")

for i, sentence in enumerate(corpus, start=1):
    print(f"{i}. {sentence}")
```

```python
# Convert text into TF-IDF numerical vectors

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)

print("TF-IDF conversion completed!")
print("Number of sentences:", tfidf_matrix.shape[0])
print("Number of unique words:", tfidf_matrix.shape[1])

print("\nTF-IDF Matrix Shape:", tfidf_matrix.shape)
```

```python
# Calculate pairwise cosine similarity

similarity_matrix = cosine_similarity(tfidf_matrix)

print("Cosine similarity matrix created successfully!")
print("\nMatrix shape:", similarity_matrix.shape)
print("\nSimilarity scores (rounded to 2 decimal places):")
print(np.round(similarity_matrix, 2))
```

```python
# Function to find the most similar sentences

def find_similar(query, corpus, top_k=3):
    # Convert the query using the same TF-IDF vectorizer
    query_vector = vectorizer.transform([query])

    # Calculate similarity between query and all corpus sentences
    scores = cosine_similarity(query_vector, tfidf_matrix)[0]

    # Get indices of the top matching sentences
    top_indices = np.argsort(scores)[::-1][:top_k]

    print(f"Query: {query}\n")
    print("Most similar sentences:\n")

    for index in top_indices:
        print(f"Score: {scores[index]:.3f} | {corpus[index]}")


print("Similarity search function created successfully!")
```

```python
# Test 1: Similar sentences
pair1 = [
    "The dog plays in the garden",
    "A puppy is a young dog"
]

# Test 2: Unrelated sentences
pair2 = [
    "The dog plays in the garden",
    "Computers process data using algorithms"
]

def compare_sentences(pair):
    vectors = vectorizer.transform(pair)
    score = cosine_similarity(vectors[0], vectors[1])[0][0]

    print("Sentence 1:", pair[0])
    print("Sentence 2:", pair[1])
    print(f"Cosine Similarity: {score:.4f}")
    print("-" * 60)

print("TEST 1: SIMILAR SENTENCES")
compare_sentences(pair1)

print("TEST 2: UNRELATED SENTENCES")
compare_sentences(pair2)
```

```python
# Visualize cosine similarity using a heatmap

plt.figure(figsize=(10, 8))

plt.imshow(similarity_matrix, cmap="YlGnBu", interpolation="nearest")
plt.colorbar(label="Cosine Similarity")

plt.xticks(range(len(corpus)), range(1, len(corpus) + 1))
plt.yticks(range(len(corpus)), range(1, len(corpus) + 1))

plt.xlabel("Sentence Number")
plt.ylabel("Sentence Number")
plt.title("TF-IDF Cosine Similarity Heatmap")

plt.tight_layout()
plt.show()
```

## Why Does Cosine Similarity Measure the Angle Instead of Distance?

Cosine similarity measures the angle between two vectors rather than the physical distance between them.

The formula is:

\[
\text{Cosine Similarity}(A,B)=\frac{A\cdot B}{\|A\|\|B\|}
\]

Where:
- \(A\cdot B\) is the dot product of the vectors.
- \(\|A\|\) and \(\|B\|\) are their magnitudes.

**Explanation:**

- A small angle indicates that the vectors point in similar directions.
- An angle of 90° gives a cosine similarity of 0.
- A larger angle indicates less similarity.

Cosine similarity focuses on the direction of vectors, making it useful for comparing text even when documents have different lengths. Euclidean distance, on the other hand, measures the straight-line distance between vectors.

**Example:** Two documents discussing similar topics may have similar word patterns even if one document is longer than the other.
