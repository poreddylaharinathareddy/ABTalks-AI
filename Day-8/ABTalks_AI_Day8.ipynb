```python

# Day 8: Text Preprocessing
!pip -q install nltk

import re
import nltk
import string

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

print("Day 8 text preprocessing tools are ready!")
```

```python

# Step 3: Create three raw text samples

social_media = "OMG!!! I LOOOVE this AI app 😍😍 it's SOOO useful!!! #AI"

news_article = "THE NEW technology is amazing... It helps people, businesses, and schools in 2026."

customer_review = "I was really disappointed :( with the service!!! The staff were not helpful, and the delivery was sooo late."

samples = {
    "Social Media": social_media,
    "News Article": news_article,
    "Customer Review": customer_review
}

for name, text in samples.items():
    print(f"\n{name}:")
    print(text)
```

```python

# Step 4: Word-level and sentence-level tokenization

for name, text in samples.items():
    print(f"\n--- {name} ---")

    words = word_tokenize(text)
    sentences = sent_tokenize(text)

    print("Word tokens:")
    print(words)

    print("\nSentence tokens:")
    print(sentences)
```

```python
from nltk.corpus import stopwords
import string

stop_words = set(stopwords.words("english"))

for name, text in samples.items():
    print(f"\n--- {name} ---")

    words_before = word_tokenize(text.lower())

    words_after = [
        word for word in words_before
        if word not in stop_words and word not in string.punctuation
    ]

    print("Before stopword removal:", words_before)
    print("After stopword removal:", words_after)
    print("Vocabulary before:", len(set(words_before)))
    print("Vocabulary after:", len(set(words_after)))
```

```python
import re
import string

for name, text in samples.items():
    cleaned_text = re.sub(r"[^\w\s]", "", text)
    cleaned_text = re.sub(r"\s+", " ", cleaned_text).strip()

    print(f"\n--- {name} ---")
    print("Before:", text)
    print("After :", cleaned_text)
```

```python
# Step 7: Compare stemming and lemmatization

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["playing", "studies", "running", "better", "cars"]

print("Word       | Stemming | Lemmatization")
print("-" * 42)

for word in words:
    stemmed = stemmer.stem(word)
    lemmatized = lemmatizer.lemmatize(word)
    print(f"{word:<10} | {stemmed:<8} | {lemmatized}")
```

```python
# Step 8: Final text preprocessing pipeline

def preprocess_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", "", text)

    # Remove punctuation, emojis, and special characters
    text = re.sub(r"[^\w\s]", "", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenize
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    return tokens


processed_dataset = []

for name, text in samples.items():
    cleaned_tokens = preprocess_text(text)

    processed_dataset.append({
        "category": name,
        "original_text": text,
        "cleaned_tokens": cleaned_tokens,
        "cleaned_text": " ".join(cleaned_tokens)
    })

final_df = pd.DataFrame(processed_dataset)

print("FINAL PREPROCESSED DATASET")
display(final_df)
```
