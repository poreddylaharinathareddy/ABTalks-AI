```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

print("NLTK setup completed successfully! 🚀")
```

```python
from nltk.tokenize import word_tokenize

text = """Artificial intelligence is changing the way we learn and work.
Python is a useful language for building AI applications."""

tokens = word_tokenize(text)

print("Original text:")
print(text)

print("\nTokens:")
print(tokens)
```

```python
from nltk.corpus import stopwords
import string

stop_words = set(stopwords.words('english'))

cleaned_tokens = [
    word.lower()
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]

print("Tokens before cleaning:")
print(tokens)

print("\nTokens after removing stop words and punctuation:")
print(cleaned_tokens)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

print("NLTK setup completed successfully! 🚀")

from nltk.tokenize import word_tokenize

text = """Artificial intelligence is changing the way we learn and work.
Python is a useful language for building AI applications."""

tokens = word_tokenize(text)

print("Original text:")
print(text)

print("\nTokens:")
print(tokens)

from nltk.corpus import stopwords
import string

stop_words = set(stopwords.words('english'))

cleaned_tokens = [
    word.lower()
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]

print("Tokens before cleaning:")
print(tokens)

print("\nTokens after removing stop words and punctuation:")
print(cleaned_tokens)

from collections import Counter

bag_of_words = Counter(cleaned_tokens)

print("Bag-of-Words representation:")
print(dict(bag_of_words))
```

```python
print("Preprocessed text dataset:")
print(cleaned_tokens)

print("\nNumber of processed tokens:", len(cleaned_tokens))
print("Number of unique words:", len(bag_of_words))
```
