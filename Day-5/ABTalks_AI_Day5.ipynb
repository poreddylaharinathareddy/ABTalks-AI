```python

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

sentences = [
    "Python is a programming language.",
    "Java is used for software development.",
    "Artificial intelligence helps machines learn.",
    "Machine learning is a part of artificial intelligence.",
    "The sun rises in the east.",
    "Deep learning uses neural networks.",
    "I enjoy learning Python programming.",
    "Dogs are friendly animals.",
    "AI can understand human language.",
    "The moon shines at night.",
    "Data science analyzes data.",
    "Neural networks are inspired by the human brain.",
    "I am studying computer science.",
    "Football is a popular sport.",
    "Python is useful for AI projects.",
    "Semantic search finds results by meaning.",
    "The ocean contains many fish.",
    "Machine learning learns patterns from data.",
    "I like building software applications.",
    "The weather is sunny today."
]

print("Dataset created successfully!")
print("Total sentences:", len(sentences))
```

```python

# Load a pre-trained sentence embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Convert all 20 sentences into numerical vectors
embeddings = model.encode(sentences)

print("Embeddings created successfully!")
print("Number of sentences:", len(embeddings))
print("Vector dimensions:", embeddings.shape)
```

```python

def semantic_search(query, top_k=3):
    # Convert the query into a vector
    query_embedding = model.encode([query])

    # Calculate similarity with all sentences
    scores = cosine_similarity(query_embedding, embeddings)[0]

    # Get the top matching sentence indexes
    top_indices = np.argsort(scores)[::-1][:top_k]

    print("Search query:", query)
    print("\nTop matching results:")

    for rank, index in enumerate(top_indices, start=1):
        print(f"{rank}. {sentences[index]}")
        print(f"   Similarity score: {scores[index]:.4f}")
```

```python

semantic_search("I want to learn AI and machine learning")

print("\n" + "=" * 50 + "\n")

semantic_search("I want to develop software using Python")

print("\n" + "=" * 50 + "\n")

semantic_search("Tell me about animals")
```

```python

def keyword_search(query):
    query_words = set(query.lower().split())
    results = []

    for sentence in sentences:
        sentence_words = set(sentence.lower().split())
        matches = len(query_words.intersection(sentence_words))

        if matches > 0:
            results.append((sentence, matches))

    results.sort(key=lambda item: item[1], reverse=True)

    print("Keyword Search Results:")
    if results:
        for sentence, score in results[:3]:
            print(f"- {sentence} (matching words: {score})")
    else:
        print("No matching sentences found.")


print("QUERY: How can machines learn?")
print("\n--- Keyword Search ---")
keyword_search("How can machines learn?")

print("\n--- Semantic Search ---")
semantic_search("How can machines learn?")
```

```python
test_queries = [
    "How do computers learn?",
    "I want to create an AI application.",
    "Tell me something about nature."
]

for query in test_queries:
    print("\n" + "=" * 60)
    semantic_search(query, top_k=3)
```

## Day 5 Reflection — Build a Mini Search Engine

Today, I built a mini semantic search engine using Python and Sentence Transformers.

I converted 20 sentences into numerical embeddings and used cosine similarity to retrieve the three most relevant results for a query.

I also compared semantic search with keyword search. I learned that keyword search matches words, while semantic search can identify related meanings even when the wording is different.

This project helped me understand how semantic search can be used in AI applications, document retrieval, and chatbots.
